```python
# 1. 필요한 패키지 설치
!pip install -q ultralytics yt-dlp opencv-python

# 2. 유튜브 영상 다운로드
import yt_dlp

# 여기에 유튜브 링크 넣기 🔽 (따옴표 안에 전체 링크 그대로!)
youtube_url = 'https://youtube.com/shorts/4im-GdrR2w8?si=3kZmYU_h3L7zEtMi'  # 👈 네 링크로 바꿔줘!

# 다운로드 옵션 설정
ydl_opts = {
    'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/mp4',
    'outtmpl': 'downloaded_video.%(ext)s',
    'merge_output_format': 'mp4'
}

# 영상 다운로드
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([youtube_url])

# 3. YOLOv8 모델 불러오기
from ultralytics import YOLO
import cv2

model = YOLO('yolov8n.pt')  # car 감지에 충분한 기본 모델

# 4. 영상 열기
cap = cv2.VideoCapture('downloaded_video.mp4')
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('output.mp4', fourcc, 20.0, (int(cap.get(3)), int(cap.get(4))))

print("자동차(car) 감지 중...")

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # YOLOv8 감지
    results = model(frame)

    # 자동차만 감지해서 로그 출력
    for result in results:
        boxes = result.boxes
        for box in boxes:
            cls = int(box.cls[0])
            label = model.names[cls]
            if label == 'car':
                print(f"프레임 {frame_count} - 'car' 감지됨! 위치:", box.xyxy[0].tolist())

    # 결과 시각화 및 저장
    annotated_frame = results[0].plot()
    out.write(annotated_frame)
    frame_count += 1

cap.release()
out.release()

print("✅ 완료! 결과 영상은 output.mp4로 저장됨.")

# 5. 다운로드 링크 제공
from google.colab import files
files.download('output.mp4')
```

```python
# 📦 1. 필요한 패키지 설치
!apt-get -y install ffmpeg
!pip install -q ultralytics yt-dlp opencv-python-headless

# 🧠 2. 유튜브 Shorts 영상 3초만 다운로드
import os
import yt_dlp

shorts_url = 'https://youtube.com/shorts/4im-GdrR2w8?si=d-MLXnOpqKWkH5jb'
video_id = shorts_url.split('/')[-1].split('?')[0]
youtube_url = f'https://www.youtube.com/watch?v={video_id}'
print("🎯 변환된 유튜브 링크:", youtube_url)

# 3초만 잘라서 저장
ydl_opts = {
    'format': 'bestvideo+bestaudio/best',
    'outtmpl': 'shorts_trimmed.mp4',
    'download_sections': ['*00:00:00-00:00:03'],
    'postprocessors': [{
        'key': 'FFmpegVideoConvertor',
        'preferedformat': 'mp4',
    }]
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([youtube_url])

assert os.path.exists('shorts_trimmed.mp4'), "❌ 다운로드된 영상 파일이 없습니다."
print("✅ 영상 다운로드 및 3초 자르기 완료")

# 🚗 3. YOLOv8으로 자동차 감지
import cv2
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
cap = cv2.VideoCapture('shorts_trimmed.mp4')
ret, frame = cap.read()
if not ret:
    raise RuntimeError("❌ 영상 열기 실패!")

height, width = frame.shape[:2]
fps = cap.get(cv2.CAP_PROP_FPS)
out = cv2.VideoWriter('output.mp4', cv2.VideoWriter_fourcc(*'mp4v'), fps, (width, height))

cap.set(cv2.CAP_PROP_POS_FRAMES, 0)
frame_count = 0
print("🚗 자동차 감지 중 (3초 영상)...")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    results = model(frame)
    annotated = results[0].plot()
    out.write(annotated)
    frame_count += 1

cap.release()
out.release()
print(f"✅ 감지 완료: output.mp4 ({frame_count} 프레임 저장됨)")

# 📥 4. 다운로드 링크 제공
from google.colab import files
import shutil

shutil.copy('output.mp4', 'yolov8_result.mp4')
print("📎 결과 영상 다운로드 링크:")
files.download('yolov8_result.mp4')
```

```python
# 📦 패키지 설치
!apt-get -y install ffmpeg
!pip install -q ultralytics yt-dlp opencv-python-headless

# 🧠 전체 실행
import os, cv2, yt_dlp, shutil
from ultralytics import YOLO
from google.colab import files

# 🔗 Shorts 링크 (앞 3초만)
shorts_url = 'https://youtube.com/shorts/4im-GdrR2w8?si=d-MLXnOpqKWkH5jb'
video_id = shorts_url.split('/')[-1].split('?')[0]
youtube_url = f'https://www.youtube.com/watch?v={video_id}'

# 🎬 yt-dlp 다운로드
ydl_opts = {
    'format': 'bestvideo+bestaudio/best',
    'outtmpl': 'shorts_trimmed.%(ext)s',
    'download_sections': ['*00:00:00-00:00:03'],
    'postprocessors': [{'key': 'FFmpegVideoConvertor', 'preferedformat': 'mp4'}]
}
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([youtube_url])

# 📁 저장된 mp4 파일 자동 탐색
video_file = next((f for f in os.listdir() if f.startswith("shorts_trimmed") and f.endswith(".mp4")), None)
if not video_file:
    raise FileNotFoundError("❌ mp4 파일을 찾을 수 없습니다.")
print(f"✅ 3초 영상 다운로드 완료 → {video_file}")

# 🚗 YOLO 감지
model = YOLO('yolov8n.pt')
cap = cv2.VideoCapture(video_file)
ret, frame = cap.read()
if not ret:
    raise RuntimeError("❌ 영상 열기 실패")
height, width = frame.shape[:2]
fps = cap.get(cv2.CAP_PROP_FPS)
out = cv2.VideoWriter("output.mp4", cv2.VideoWriter_fourcc(*"mp4v"), fps, (width, height))
cap.set(cv2.CAP_PROP_POS_FRAMES, 0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    results = model(frame)
    out.write(results[0].plot())

cap.release()
out.release()
print("✅ 감지 완료: output.mp4 저장됨")

# 📥 다운로드 링크 제공
shutil.copy("output.mp4", "yolov8_result.mp4")
files.download("yolov8_result.mp4")
```
